```python
# Regular class
from itertools import product



# Simulation parameters class
class SimParam:
    def __init__(self, sim, vrest, gain):
        self.sim = sim
        self.vrest = vrest
        self.gain = gain

    def __repr__(self):
        return f"SimParam(sim={self.sim}, vrest={self.vrest}, gain={self.gain})"

# Configuration (sim * pqif * seed) where simulation parameters are defined by the above class
class Config:
    def __init__(self, name, sims, pqif_numbers, seed_numbers):
        self.name = name
        self.sims = sims
        self.pqif_numbers = pqif_numbers
        self.seed_numbers = seed_numbers

    def iter_params(self):
        '''Creates a product of the iteration variables'''

        for sim_param in self.sims:
            for pqif in self.pqif_numbers:
                for seed in self.seed_numbers:
                    yield sim_param, pqif, seed  # gives the combo


# Defining simulation groups
oscillation_sims = [
    SimParam(1, -8.5, 14.44),
    SimParam(2, -12.3, 10.68),
    SimParam(3, -17.0, 8.65),
    SimParam(4, -22.0, 7.18),
]

sequence_sims = [
    SimParam(9, -8.5, 14.44),
    SimParam(10, -12.3, 10.68),
    SimParam(11, -17.0, 8.65),
    SimParam(12, -22.0, 7.18),
]

sequence_param = [
    SimParam(1, -8.5, 14.44),
    SimParam(2, -12.3, 10.68),
    SimParam(3, -17.0, 8.65),
    SimParam(4, -22.0, 7.18),
]

# Shared parameters
all_pqif = [0, 0.25, 0.5, 0.75, 1]
homogeneous_pqif = [0, 1]
seeds = range(10)


# Creates class instances in various configurations that we are interested in. Uses nested class for the simulation parameter
configs = [
    Config(
        name="oscillations_heterogeneity",
        sims=oscillation_sims,
        pqif_numbers=all_pqif,
        seed_numbers=seeds,
    ),
    Config(
        name="sequences_heterogeneity",
        sims=sequence_sims,
        pqif_numbers=all_pqif,
        seed_numbers=seeds,
    ),
    Config(
        name="oscillations_homogeneous",
        sims=oscillation_sims,
        pqif_numbers=homogeneous_pqif,
        seed_numbers=seeds,
    ),
]

# Example usage:

print("LIST OF CONFIGURATIONS:")
for i, cfg in enumerate(configs):
    print(f"{i}: {cfg.name}")

for sim_param, pqif, seed in cfg.iter_params():
    sim = sim_param.sim
    vrest = sim_param.vrest
    gain = sim_param.gain

    print(cfg.name, sim, vrest, gain, pqif, seed)

    # load / analyze / plot here
```
